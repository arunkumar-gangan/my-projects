# 6. Hollow Square Pattern

```python
n = 5
for i in range(n):
    for j in range(n):
        if i == 0 or i == n - 1 or j == 0 or j == n - 1:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

# 9. Right-Aligned Triangle

```python
n = 5
for i in range(1, n + 1):
    print(" " * (n - i) + "*" * i)
```

# 10. Diamond Pattern

```python
n = 5
# Upper part
for i in range(n):
    print(" " * (n - i - 1) + "* " * (i + 1))
# Lower part
for i in range(n - 2, -1, -1):
    print(" " * (n - i - 1) + "* " * (i + 1))
```

# 11. Floyd’s Triangle

```python
n = 5
num = 1
for i in range(1, n + 1):
    for j in range(i):
        print(num, end=" ")
        num += 1
    print()
```

# 12. Pascal’s Triangle

```python
n = 5
for i in range(n):
    num = 1
    print(" " * (n - i), end="")
    for j in range(i + 1):
        print(num, end=" ")
        num = num * (i - j) // (j + 1)
    print()
```

# 13. Alternating 0 and 1 Triangle

```python
n = 7
for i in range(n):
    for j in range(i + 1):
        print((i + j) % 2, end=" ")
    print()
```

# 14. Continuous Number Triangle

```python
n = 5
num = 1
for i in range(1, n + 1):
    for j in range(i):
        print(num, end=" ")
        num += 1
    print()
```

# 15. Star X Pattern

```python
n = 5
for i in range(n):
    for j in range(n):
        if i == j or j == n - i - 1:
            print("*", end="")
        else:
            print(" ", end="")
    print()
```

# 16. Hollow Right-Angled Triangle

```python
n = 5
for i in range(1, n + 1):
    for j in range(1, i + 1):
        if j == 1 or j == i or i == n:
            print("*", end="")
        else:
            print(" ", end="")
    print()
```
